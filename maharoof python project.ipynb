## STEP 5

```python
plt.figure(figsize=(7,5))
sns.countplot(data=card,x="card")
plt.title("credit card approval distribution")
plt.xlabel("card approval")
plt.ylabel("Number of applications")
plt.show()
```

>It shows the number of approved and rejected applications 

```python
plt.figure(figsize=(8,5))
sns.histplot(card["age"],bins=20,kde=True)
plt.title("Age distribution of applications")
plt.xlabel("Age")
plt.ylabel("frequency")
plt.show()
```

>It shows the age distribution of applications 

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=card,x="card",y="income")
plt.title("income vs credit card approval")
plt.xlabel("card approval")
plt.ylabel("income")
plt.show()
```

>It shows income levels of approved and rejected applicants

```python
plt.figure(figsize=(10,5))
sns.histplot(card["expenditure"],bins=30,kde=True)
plt.title("Customer expenditure distribution")
plt.xlabel("expenditure")
plt.ylabel("frequency")
plt.show()
```

>It shows the customer expenditure patterns

```python
plt.figure(figsize=(10,5))
sns.countplot(data=card,x="owner",hue="card")
plt.title("Home ownership vs card approval")
plt.xlabel("Home ownership")
plt.ylabel("Number of applicants")
plt.legend(title="Card approval")
plt.show()
```

>It compares the credit card approval among homeowners and non-homeowners

## step 6 - insights and conclusion

1. Credit card approval - 
Out of 1319 applications,1023 applications were approved and 296 were rejected.Approximately 77.56% of
applications were approved

 2.  Customer Age - The average age of applicants is approximately 33.21 years .This indicates that the dataset mainly contains
 adult applicants

 3. Customer income - The average is approximately 3.37 in the dataset's recorded units.Approved applicants have a higher average 
income (3.45)than rejected applicants(3.07).

 4. Major Credit Cards
Applicants who already had a major credit card had a higher approval rate (79.68%) compared to those who did not (68.05%).

 5. Active Credit Accounts
Approved applicants had an average of 7.27 active credit accounts, while rejected applicants had an average of 6.05.
